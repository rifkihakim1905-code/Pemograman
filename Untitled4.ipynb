```python
import seaborn as sns
import matplotlib.pyplot as plt

# data
Bulan = ['Jan', 'Feb', 'Mar', 'April', 'Mei', 'Juni', 'Juli']
Belanja= [ 100000, 200000, 130000, 250000, 129000, 230000, 167000]

plt.figure(figsize=(8, 4))

sns.barplot(x=Bulan, y=Belanja, palette='pastel')

plt.plot(
    range(len(Bulan)),
    Belanja,
    color='black',
    marker='o',
    linestyle='-',
    linewidth=2,
    label='Garis Tren'
)

plt.title('Belanja Per Bulan Dengan Garis', fontsize=11, fontweight='bold')
plt.xlabel('Bulan')
plt.ylabel('Jumlah Balanja')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Muat dataset
url = "https://raw.githubusercontent.com/Alam-A99/Pandas_Latih/10ebb3af000e930a44989075733f495d5c2f4be5/penjualan.csv"
df = pd.read_csv(url)

# 2. Tambahkan kolom total untuk memudahkan beberapa grafik
df['total'] = df[['jan', 'feb', 'mar', 'apr']].sum(axis=1)

# Atur style seaborn agar lebih rapi
sns.set_theme(style="whitegrid")

# Bar Chart
plt.figure(figsize=(8, 5))

# Urutkan data dari total tertinggi
df_sorted = df.sort_values('total', ascending=False)

sns.barplot(data=df_sorted, x='nama', y='total', palette='viridis')
plt.title('Total Penjualan per Orang', fontsize=12)
plt.xlabel('Nama Karyawan')
plt.ylabel('Total Penjualan')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Pie Chart
plt.figure(figsize=(5, 5)) # Ukuran persegi agar pie chart tidak gepeng

# Hitung total penjualan per bulan
total_per_bulan = df[['jan', 'feb', 'mar', 'apr']].sum()

plt.pie(total_per_bulan, labels=total_per_bulan.index, autopct='%1.1f%%',
        startangle=140, colors=sns.color_palette('muted'))

plt.title('Proporsi Penjualan per Bulan', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import plotly.express as px

# Ubah format data menjadi 'long' (khusus mengambil kolom 'jan' dan 'mar')
df_melted = df.melt(id_vars='nama', value_vars=['jan', 'mar'], var_name='Bulan', value_name='Penjualan')

# Buat Line Chart (Hanya butuh 1 blok kode inti!)
fig = px.line(
    df_melted, x='nama', y='Penjualan', color='Bulan', symbol='Bulan', symbol_sequence=['circle', 'triangle-up'], color_discrete_sequence=['royalblue', 'crimson'],  markers=True,
    title='Perbandingan Penjualan: Januari vs Maret',
    labels={'nama': 'Nama Karyawan', 'Penjualan': 'Jumlah Penjualan'} # Ubah label sumbu
)

# Atur ukuran agar tidak full screen (Opsional)
fig.update_layout(width=800, height=500)

# 5. Tampilkan grafik
fig.show()
```

```python
# Boxplot (Sebaran Kemiskinan & Outlier)
plt.figure(figsize=(6, 5))
sns.boxplot(data=df1, y='KEMISKINAN', color='salmon', width=0.4)

plt.title('Sebaran Tingkat Kemiskinan', fontsize=12, fontweight='bold')
plt.ylabel('Persentase Kemiskinan (%)')
plt.tight_layout()
plt.show()
```

```python
# Boxplot (Sebaran IPM & Outlier)
# Melihat median, kuartil, dan daerah mana yang menjadi outlier (pencilan) dalam IPM.
plt.figure(figsize=(6, 5))
sns.boxplot(data=df1, y='IPM', color='skyblue', width=0.4)

plt.title('Sebaran Indeks Pembangunan Manusia (IPM)', fontsize=12, fontweight='bold')
plt.ylabel('Nilai IPM (0-100)')
plt.tight_layout()
plt.show()
```
